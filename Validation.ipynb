```python
# ============================================================
# STEP 12 — SINGLE VALIDATION PREDICTION
# ============================================================

import torch

print("=" * 70)
print("GENERATING VALIDATION PREDICTION")
print("=" * 70)

# ------------------------------------------------------------
# Use the SAME messages structure stored in validation JSONL
# ------------------------------------------------------------

messages = record["messages"].copy()

# We only want the user message when generating.
user_messages = [
    messages[0]
]

# Add generation prompt
prompt = processor.apply_chat_template(
    user_messages,
    tokenize=False,
    add_generation_prompt=True
)

print("\nPrompt:")
print(prompt)


# ------------------------------------------------------------
# Processor
# ------------------------------------------------------------

inputs = processor(
    text=[prompt],
    images=[image],
    return_tensors="pt"
)

device = next(
    eval_model.parameters()
).device

for key in inputs:

    if torch.is_tensor(inputs[key]):

        inputs[key] = inputs[key].to(
            device
        )


print("\nInput shapes:")

for key, value in inputs.items():

    if torch.is_tensor(value):

        print(
            f"{key:<25}: {tuple(value.shape)}"
        )


# ------------------------------------------------------------
# Generate
# ------------------------------------------------------------

print("\nGenerating...")

with torch.inference_mode():

    generated_ids = eval_model.generate(
        **inputs,
        max_new_tokens=20,
        do_sample=False
    )


# ------------------------------------------------------------
# Decode ONLY generated tokens
# ------------------------------------------------------------

input_length = (
    inputs["input_ids"].shape[1]
)

answer_ids = generated_ids[
    0,
    input_length:
]

prediction = processor.decode(
    answer_ids,
    skip_special_tokens=True
).strip()


print("\n" + "=" * 70)
print("VALIDATION RESULT")
print("=" * 70)

print("Patch ID:")
print(record["patch_id"])

print("\nQuestion:")
print(question)

print("\nGround truth:")
print(ground_truth)

print("\nModel prediction:")
print(prediction)

print("=" * 70)
```

```python
# ============================================================
# STEP 13 — RESUMABLE FULL VALIDATION EVALUATION
# ============================================================

import os
import json
import re
import time
import torch

from PIL import Image
from tqdm.auto import tqdm
from collections import defaultdict


# ============================================================
# OUTPUT FILES
# ============================================================

EVAL_DIR = os.path.join(
    PROJECT_DIR,
    "evaluation_results"
)

os.makedirs(
    EVAL_DIR,
    exist_ok=True
)

RESULTS_FILE = os.path.join(
    EVAL_DIR,
    "validation_predictions.jsonl"
)

SUMMARY_FILE = os.path.join(
    EVAL_DIR,
    "validation_summary.json"
)

INCORRECT_FILE = os.path.join(
    EVAL_DIR,
    "validation_incorrect.jsonl"
)

print("=" * 70)
print("SATQUERY FULL VALIDATION EVALUATION")
print("=" * 70)

print("Validation examples :", len(val_records))
print("Results file        :", RESULTS_FILE)
print("Summary file        :", SUMMARY_FILE)
print("Incorrect examples  :", INCORRECT_FILE)


# ============================================================
# ANSWER NORMALIZATION
# ============================================================

def normalize_answer(answer):

    if answer is None:
        return ""

    answer = str(answer).strip().lower()

    answer = re.sub(
        r"[.,!?;:\"'()\[\]{}]",
        " ",
        answer
    )

    answer = re.sub(
        r"\s+",
        " ",
        answer
    )

    return answer.strip()


def answers_match(
    prediction,
    ground_truth
):

    p = normalize_answer(prediction)
    g = normalize_answer(ground_truth)

    # ----------------------------------------
    # Exact match
    # ----------------------------------------

    if p == g:
        return True

    # ----------------------------------------
    # Yes / No
    # ----------------------------------------

    if g in ["yes", "no"]:

        words = p.split()

        if len(words) > 0:

            return words[0] == g

    # ----------------------------------------
    # Multiple choice
    # ----------------------------------------

    if g in ["a", "b", "c", "d"]:

        if p.startswith(g):
            return True

        match = re.match(
            r"^([abcd])[\s\)\.\:\-]",
            p
        )

        if match:

            return (
                match.group(1) == g
            )

    return False


# ============================================================
# RESUME FROM EXISTING RESULTS
# ============================================================

completed_indices = set()

if os.path.isfile(RESULTS_FILE):

    print("\nExisting results found.")

    with open(
        RESULTS_FILE,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            if not line.strip():
                continue

            try:

                result = json.loads(line)

                completed_indices.add(
                    result["index"]
                )

            except Exception:

                continue

    print(
        "Already evaluated:",
        len(completed_indices)
    )

else:

    print("\nNo previous evaluation found.")
    print("Starting from validation example 0.")


remaining = [
    i
    for i in range(len(val_records))
    if i not in completed_indices
]

print(
    "Remaining examples:",
    len(remaining)
)


# ============================================================
# IF EVERYTHING IS DONE
# ============================================================

if len(remaining) == 0:

    print("\n✅ All validation examples already evaluated.")

else:

    # --------------------------------------------------------
    # Statistics for current run
    # --------------------------------------------------------

    run_correct = 0
    run_evaluated = 0
    run_failed = 0

    start_time = time.time()


    # --------------------------------------------------------
    # Open result files
    # --------------------------------------------------------

    with open(
        RESULTS_FILE,
        "a",
        encoding="utf-8"
    ) as result_file, open(
        INCORRECT_FILE,
        "a",
        encoding="utf-8"
    ) as incorrect_file:


        # ----------------------------------------------------
        # Progress bar
        # ----------------------------------------------------

        progress = tqdm(
            remaining,
            desc="Evaluating validation",
            unit="example"
        )


        for index in progress:

            record = val_records[index]

            try:

                # ==================================================
                # EXTRACT FROM ACTUAL VALIDATION SCHEMA
                # ==================================================

                user_message = (
                    record["messages"][0]
                )

                content = (
                    user_message["content"]
                )


                # --------------------------------------------------
                # Find image and text safely
                # --------------------------------------------------

                image_relative_path = None
                question = None

                for item in content:

                    if item.get("type") == "image":

                        image_relative_path = (
                            item["image"]
                        )

                    elif item.get("type") == "text":

                        question = (
                            item["text"]
                        )


                if image_relative_path is None:

                    raise ValueError(
                        "Image path not found in messages"
                    )


                if question is None:

                    raise ValueError(
                        "Question not found in messages"
                    )


                # ==================================================
                # GROUND TRUTH
                # ==================================================

                ground_truth = (
                    record["messages"]
                    [1]
                    ["content"]
                    [0]
                    ["text"]
                )


                # ==================================================
                # IMAGE
                # ==================================================

                image_path = os.path.join(
                    DATASET_DIR,
                    image_relative_path
                )


                if not os.path.isfile(
                    image_path
                ):

                    raise FileNotFoundError(
                        image_path
                    )


                image = Image.open(
                    image_path
                ).convert("RGB")


                # ==================================================
                # USE EXACT VALIDATION USER MESSAGE
                # ==================================================

                messages = [
                    user_message
                ]


                prompt = (
                    processor.apply_chat_template(
                        messages,
                        tokenize=False,
                        add_generation_prompt=True
                    )
                )


                # ==================================================
                # PROCESS IMAGE + TEXT
                # ==================================================

                inputs = processor(
                    text=[prompt],
                    images=[image],
                    return_tensors="pt"
                )


                device = next(
                    eval_model.parameters()
                ).device


                for key in inputs:

                    if torch.is_tensor(
                        inputs[key]
                    ):

                        inputs[key] = (
                            inputs[key].to(device)
                        )


                # ==================================================
                # GENERATE
                # ==================================================

                with torch.inference_mode():

                    generated_ids = (
                        eval_model.generate(
                            **inputs,
                            max_new_tokens=20,
                            do_sample=False
                        )
                    )


                # ==================================================
                # ONLY DECODE NEW TOKENS
                # ==================================================

                input_length = (
                    inputs["input_ids"].shape[1]
                )

                answer_ids = (
                    generated_ids[
                        0,
                        input_length:
                    ]
                )


                prediction = processor.decode(
                    answer_ids,
                    skip_special_tokens=True
                ).strip()


                # ==================================================
                # CHECK ANSWER
                # ==================================================

                is_correct = answers_match(
                    prediction,
                    ground_truth
                )


                # ==================================================
                # SAVE RESULT IMMEDIATELY
                # ==================================================

                output_record = {

                    "index": index,

                    "patch_id":
                        record["patch_id"],

                    "question":
                        question,

                    "ground_truth":
                        ground_truth,

                    "prediction":
                        prediction,

                    "correct":
                        is_correct,

                    "type":
                        record.get(
                            "type",
                            "unknown"
                        ),

                    "category":
                        record.get(
                            "category",
                            "unknown"
                        ),

                    "image":
                        image_relative_path

                }


                result_file.write(
                    json.dumps(
                        output_record,
                        ensure_ascii=False
                    )
                    + "\n"
                )

                result_file.flush()


                # --------------------------------------------------
                # Save incorrect examples separately
                # --------------------------------------------------

                if not is_correct:

                    incorrect_file.write(
                        json.dumps(
                            output_record,
                            ensure_ascii=False
                        )
                        + "\n"
                    )

                    incorrect_file.flush()


                # ==================================================
                # UPDATE STATISTICS
                # ==================================================

                run_evaluated += 1

                if is_correct:

                    run_correct += 1


                run_accuracy = (
                    run_correct
                    /
                    run_evaluated
                    *
                    100
                    if run_evaluated
                    else 0
                )


                progress.set_postfix(
                    accuracy=f"{run_accuracy:.2f}%"
                )


                # ==================================================
                # CLEANUP
                # ==================================================

                del inputs
                del generated_ids
                del answer_ids
                del image

                if torch.cuda.is_available():

                    torch.cuda.empty_cache()


            except Exception as e:

                run_failed += 1

                print(
                    f"\n⚠️ FAILED INDEX: {index}"
                )

                print(
                    "Patch:",
                    record.get(
                        "patch_id",
                        "unknown"
                    )
                )

                print(
                    "Error:",
                    repr(e)
                )

                continue


    elapsed = (
        time.time()
        -
        start_time
    )


    print("\n" + "=" * 70)
    print("CURRENT EVALUATION RUN COMPLETE")
    print("=" * 70)

    print(
        "Evaluated this run:",
        run_evaluated
    )

    print(
        "Correct:",
        run_correct
    )

    print(
        "Failed:",
        run_failed
    )

    print(
        "Run accuracy:",
        f"{run_correct / run_evaluated * 100:.2f}%"
        if run_evaluated
        else "N/A"
    )

    print(
        "Time:",
        f"{elapsed / 60:.1f} minutes"
    )

    print("=" * 70)
```

```python
# ============================================================
# STEP 13 — SAFE RESUMABLE VALIDATION EVALUATION
# ============================================================

import os
import json
import re
import time
import torch

from PIL import Image
from tqdm.auto import tqdm


# ============================================================
# PATHS
# ============================================================

EVAL_DIR = os.path.join(
    PROJECT_DIR,
    "evaluation_results"
)

os.makedirs(
    EVAL_DIR,
    exist_ok=True
)

RESULTS_FILE = os.path.join(
    EVAL_DIR,
    "validation_predictions.jsonl"
)

PROGRESS_FILE = os.path.join(
    EVAL_DIR,
    "evaluation_progress.json"
)

INCORRECT_FILE = os.path.join(
    EVAL_DIR,
    "validation_incorrect.jsonl"
)


print("=" * 70)
print("SATQUERY — SAFE VALIDATION EVALUATION")
print("=" * 70)

print("Validation examples :", len(val_records))
print("Results             :", RESULTS_FILE)
print("Progress            :", PROGRESS_FILE)
print("Incorrect           :", INCORRECT_FILE)


# ============================================================
# NORMALIZATION
# ============================================================

def normalize_answer(answer):

    if answer is None:
        return ""

    answer = str(answer).strip().lower()

    answer = re.sub(
        r"[.,!?;:\"'()\[\]{}]",
        " ",
        answer
    )

    answer = re.sub(
        r"\s+",
        " ",
        answer
    )

    return answer.strip()


def answers_match(
    prediction,
    ground_truth
):

    p = normalize_answer(prediction)
    g = normalize_answer(ground_truth)

    # Exact match
    if p == g:
        return True

    # Yes / No questions
    if g in ["yes", "no"]:

        if p.startswith(g):
            return True

    # Multiple choice
    if g in ["a", "b", "c", "d"]:

        if p.startswith(g):
            return True

    return False


# ============================================================
# DETERMINE COMPLETED EXAMPLES
# ============================================================

completed_indices = set()

if os.path.exists(RESULTS_FILE):

    print("\nReading existing predictions...")

    with open(
        RESULTS_FILE,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            if not line.strip():
                continue

            try:

                result = json.loads(line)

                if "index" in result:

                    completed_indices.add(
                        int(result["index"])
                    )

            except Exception:

                pass


print(
    "Previously completed:",
    len(completed_indices)
)


remaining_indices = [
    i
    for i in range(len(val_records))
    if i not in completed_indices
]


print(
    "Remaining:",
    len(remaining_indices)
)


# ============================================================
# OPEN FILES
# ============================================================

result_file = open(
    RESULTS_FILE,
    "a",
    encoding="utf-8",
    buffering=1
)

incorrect_file = open(
    INCORRECT_FILE,
    "a",
    encoding="utf-8",
    buffering=1
)


# ============================================================
# STATISTICS
# ============================================================

evaluated = 0
correct = 0
failed = 0

start_time = time.time()


# ============================================================
# EVALUATION LOOP
# ============================================================

try:

    for index in tqdm(
        remaining_indices,
        desc="Evaluating validation",
        unit="example"
    ):

        record = val_records[index]

        try:

            # ====================================================
            # EXTRACT USER MESSAGE
            # ====================================================

            user_message = record["messages"][0]

            content = user_message["content"]

            image_relative_path = None
            question = None

            for item in content:

                if item.get("type") == "image":

                    image_relative_path = item["image"]

                elif item.get("type") == "text":

                    question = item["text"]


            if image_relative_path is None:

                raise ValueError(
                    "Image path missing"
                )


            if question is None:

                raise ValueError(
                    "Question missing"
                )


            # ====================================================
            # GROUND TRUTH
            # ====================================================

            ground_truth = (
                record["messages"]
                [1]
                ["content"]
                [0]
                ["text"]
            )


            # ====================================================
            # IMAGE PATH
            # ====================================================

            image_path = os.path.join(
                DATASET_DIR,
                image_relative_path
            )


            if not os.path.isfile(image_path):

                raise FileNotFoundError(
                    image_path
                )


            # ====================================================
            # LOAD RGB IMAGE
            # ====================================================

            image = Image.open(
                image_path
            ).convert("RGB")


            # ====================================================
            # QWEN PROMPT
            # ====================================================

            prompt = processor.apply_chat_template(
                [user_message],
                tokenize=False,
                add_generation_prompt=True
            )


            # ====================================================
            # PROCESS
            # ====================================================

            inputs = processor(
                text=[prompt],
                images=[image],
                return_tensors="pt"
            )


            device = next(
                eval_model.parameters()
            ).device


            for key in inputs:

                if torch.is_tensor(
                    inputs[key]
                ):

                    inputs[key] = inputs[key].to(
                        device
                    )


            # ====================================================
            # GENERATE
            # ====================================================

            with torch.inference_mode():

                generated_ids = eval_model.generate(
                    **inputs,
                    max_new_tokens=20,
                    do_sample=False
                )


            # ====================================================
            # REMOVE INPUT TOKENS
            # ====================================================

            input_length = (
                inputs["input_ids"].shape[1]
            )

            answer_ids = generated_ids[
                0,
                input_length:
            ]


            prediction = processor.decode(
                answer_ids,
                skip_special_tokens=True
            ).strip()


            # ====================================================
            # CHECK
            # ====================================================

            is_correct = answers_match(
                prediction,
                ground_truth
            )


            # ====================================================
            # RESULT
            # ====================================================

            result = {

                "index": index,

                "patch_id":
                    record.get(
                        "patch_id"
                    ),

                "question":
                    question,

                "ground_truth":
                    ground_truth,

                "prediction":
                    prediction,

                "correct":
                    is_correct,

                "type":
                    record.get(
                        "type",
                        "unknown"
                    ),

                "category":
                    record.get(
                        "category",
                        "unknown"
                    ),

                "image":
                    image_relative_path
            }


            # ====================================================
            # SAVE IMMEDIATELY
            # ====================================================

            result_file.write(
                json.dumps(
                    result,
                    ensure_ascii=False
                )
                + "\n"
            )

            result_file.flush()


            # ====================================================
            # INCORRECT FILE
            # ====================================================

            if not is_correct:

                incorrect_file.write(
                    json.dumps(
                        result,
                        ensure_ascii=False
                    )
                    + "\n"
                )

                incorrect_file.flush()


            # ====================================================
            # UPDATE STATS
            # ====================================================

            evaluated += 1

            if is_correct:

                correct += 1


            # ====================================================
            # EVERY 10 EXAMPLES:
            # FORCE DRIVE FILE SYNC + PROGRESS FILE
            # ====================================================

            if evaluated % 10 == 0:

                os.fsync(
                    result_file.fileno()
                )

                os.fsync(
                    incorrect_file.fileno()
                )

                progress_data = {

                    "total_validation_examples":
                        len(val_records),

                    "completed_total":
                        len(completed_indices)
                        + evaluated,

                    "evaluated_this_run":
                        evaluated,

                    "correct_this_run":
                        correct,

                    "failed_this_run":
                        failed,

                    "last_index":
                        index,

                    "timestamp":
                        time.time()
                }


                temp_progress = (
                    PROGRESS_FILE
                    + ".tmp"
                )


                with open(
                    temp_progress,
                    "w",
                    encoding="utf-8"
                ) as pf:

                    json.dump(
                        progress_data,
                        pf,
                        indent=2
                    )

                    pf.flush()

                    os.fsync(
                        pf.fileno()
                    )


                os.replace(
                    temp_progress,
                    PROGRESS_FILE
                )


            # ====================================================
            # CLEANUP
            # ====================================================

            del inputs
            del generated_ids
            del answer_ids
            del image

            if torch.cuda.is_available():

                torch.cuda.empty_cache()


        except Exception as e:

            failed += 1

            print(
                f"\n⚠️ FAILED INDEX {index}"
            )

            print(
                "Patch:",
                record.get(
                    "patch_id",
                    "unknown"
                )
            )

            print(
                "Error:",
                repr(e)
            )

            # Save failure too
            failure_record = {

                "index": index,

                "patch_id":
                    record.get(
                        "patch_id"
                    ),

                "error":
                    repr(e),

                "failed":
                    True
            }

            result_file.write(
                json.dumps(
                    failure_record
                )
                + "\n"
            )

            result_file.flush()

            continue


finally:

    # ============================================================
    # FINAL SYNC
    # ============================================================

    result_file.flush()
    incorrect_file.flush()

    os.fsync(
        result_file.fileno()
    )

    os.fsync(
        incorrect_file.fileno()
    )

    result_file.close()
    incorrect_file.close()


# ============================================================
# FINAL PROGRESS SAVE
# ============================================================

elapsed = (
    time.time()
    -
    start_time
)

final_progress = {

    "total_validation_examples":
        len(val_records),

    "completed_total":
        len(completed_indices)
        + evaluated,

    "evaluated_this_run":
        evaluated,

    "correct_this_run":
        correct,

    "failed_this_run":
        failed,

    "finished":
        (
            len(completed_indices)
            + evaluated
            >= len(val_records)
        ),

    "elapsed_seconds":
        elapsed,

    "timestamp":
        time.time()
}


with open(
    PROGRESS_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_progress,
        f,
        indent=2
    )

    f.flush()

    os.fsync(
        f.fileno()
    )


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("VALIDATION RUN FINISHED")
print("=" * 70)

print(
    "Evaluated this run:",
    evaluated
)

print(
    "Correct:",
    correct
)

print(
    "Failed:",
    failed
)

if evaluated > 0:

    print(
        "Accuracy this run:",
        f"{correct / evaluated * 100:.2f}%"
    )

print(
    "Time:",
    f"{elapsed / 60:.2f} minutes"
)

print(
    "\nResults saved to:",
    RESULTS_FILE
)

print(
    "Progress saved to:",
    PROGRESS_FILE
)

print(
    "Incorrect answers:",
    INCORRECT_FILE
)

print("=" * 70)
```

```python
# ============================================================
# SATQUERY — ANALYZE SAVED VALIDATION RESULTS
# NO MODEL / NO GPU / NO TRAINING REQUIRED
# ============================================================

import os
import json
from collections import Counter, defaultdict

EVAL_DIR = "/content/drive/MyDrive/finalProject/evaluation_results"

RESULTS_FILE = os.path.join(
    EVAL_DIR,
    "validation_predictions.jsonl"
)

SUMMARY_FILE = os.path.join(
    EVAL_DIR,
    "saved_results_summary.json"
)

print("=" * 70)
print("SATQUERY — SAVED VALIDATION RESULTS ANALYSIS")
print("=" * 70)

if not os.path.exists(RESULTS_FILE):
    raise FileNotFoundError(
        f"Results file not found:\n{RESULTS_FILE}"
    )

# ============================================================
# READ SAVED RESULTS
# ============================================================

results = []

with open(
    RESULTS_FILE,
    "r",
    encoding="utf-8"
) as f:

    for line in f:

        if not line.strip():
            continue

        try:
            results.append(
                json.loads(line)
            )
        except json.JSONDecodeError:
            pass


print(
    "\nSaved records:",
    len(results)
)

# ============================================================
# REMOVE FAILED RECORDS FROM ACCURACY
# ============================================================

valid_results = [
    r for r in results
    if not r.get("failed", False)
    and "correct" in r
]

failed_results = [
    r for r in results
    if r.get("failed", False)
]

print(
    "Valid predictions:",
    len(valid_results)
)

print(
    "Failed predictions:",
    len(failed_results)
)

# ============================================================
# OVERALL ACCURACY
# ============================================================

correct = sum(
    1
    for r in valid_results
    if r.get("correct") is True
)

incorrect = sum(
    1
    for r in valid_results
    if r.get("correct") is False
)

if len(valid_results) > 0:

    accuracy = (
        correct /
        len(valid_results)
        * 100
    )

else:

    accuracy = 0


print("\n" + "=" * 70)
print("OVERALL RESULTS")
print("=" * 70)

print(
    f"Correct predictions : {correct}"
)

print(
    f"Incorrect predictions: {incorrect}"
)

print(
    f"Valid predictions   : {len(valid_results)}"
)

print(
    f"Accuracy            : {accuracy:.2f}%"
)

# ============================================================
# ACCURACY BY QUESTION TYPE
# ============================================================

type_stats = defaultdict(
    lambda: {
        "total": 0,
        "correct": 0
    }
)

for r in valid_results:

    qtype = r.get(
        "type",
        "unknown"
    )

    type_stats[qtype]["total"] += 1

    if r.get("correct"):

        type_stats[qtype]["correct"] += 1


print("\n" + "=" * 70)
print("ACCURACY BY QUESTION TYPE")
print("=" * 70)

for qtype, stats in sorted(
    type_stats.items()
):

    total = stats["total"]
    corr = stats["correct"]

    acc = (
        corr / total * 100
        if total
        else 0
    )

    print(
        f"{qtype:<20}"
        f" Total: {total:<6}"
        f" Correct: {corr:<6}"
        f" Accuracy: {acc:.2f}%"
    )

# ============================================================
# ACCURACY BY CATEGORY
# ============================================================

category_stats = defaultdict(
    lambda: {
        "total": 0,
        "correct": 0
    }
)

for r in valid_results:

    category = r.get(
        "category",
        "unknown"
    )

    category_stats[
        category
    ]["total"] += 1

    if r.get("correct"):

        category_stats[
            category
        ]["correct"] += 1


print("\n" + "=" * 70)
print("ACCURACY BY CATEGORY")
print("=" * 70)

for category, stats in sorted(
    category_stats.items()
):

    total = stats["total"]
    corr = stats["correct"]

    acc = (
        corr / total * 100
        if total
        else 0
    )

    print(
        f"{category:<25}"
        f" Total: {total:<6}"
        f" Correct: {corr:<6}"
        f" Accuracy: {acc:.2f}%"
    )

# ============================================================
# GROUND TRUTH DISTRIBUTION
# ============================================================

ground_truth_counter = Counter(
    str(
        r.get(
            "ground_truth",
            ""
        )
    ).strip().lower()
    for r in valid_results
)

prediction_counter = Counter(
    str(
        r.get(
            "prediction",
            ""
        )
    ).strip().lower()
    for r in valid_results
)

print("\n" + "=" * 70)
print("GROUND TRUTH DISTRIBUTION")
print("=" * 70)

for answer, count in ground_truth_counter.most_common():

    print(
        f"{answer:<20}: {count}"
    )

print("\n" + "=" * 70)
print("PREDICTION DISTRIBUTION")
print("=" * 70)

for answer, count in prediction_counter.most_common():

    print(
        f"{answer:<20}: {count}"
    )

# ============================================================
# INCORRECT EXAMPLES
# ============================================================

incorrect_examples = [
    r
    for r in valid_results
    if not r.get("correct")
]

print("\n" + "=" * 70)
print("INCORRECT EXAMPLES")
print("=" * 70)

print(
    "Total incorrect:",
    len(incorrect_examples)
)

for r in incorrect_examples[:20]:

    print("\n----------------------------------------")

    print(
        "Index:",
        r.get("index")
    )

    print(
        "Patch:",
        r.get("patch_id")
    )

    print(
        "Type:",
        r.get("type")
    )

    print(
        "Category:",
        r.get("category")
    )

    print(
        "Question:",
        r.get("question")
    )

    print(
        "Ground truth:",
        r.get("ground_truth")
    )

    print(
        "Prediction:",
        r.get("prediction")
    )

# ============================================================
# SAVE SUMMARY
# ============================================================

summary = {

    "results_file":
        RESULTS_FILE,

    "saved_records":
        len(results),

    "valid_predictions":
        len(valid_results),

    "failed_predictions":
        len(failed_results),

    "correct":
        correct,

    "incorrect":
        incorrect,

    "accuracy_percent":
        round(
            accuracy,
            4
        ),

    "accuracy_by_type": {},

    "accuracy_by_category": {},

    "ground_truth_distribution":
        dict(
            ground_truth_counter
        ),

    "prediction_distribution":
        dict(
            prediction_counter
        )
}


for qtype, stats in type_stats.items():

    total = stats["total"]
    corr = stats["correct"]

    summary[
        "accuracy_by_type"
    ][qtype] = {

        "total": total,

        "correct": corr,

        "accuracy_percent":
            round(
                corr / total * 100,
                4
            )
            if total
            else 0
    }


for category, stats in category_stats.items():

    total = stats["total"]
    corr = stats["correct"]

    summary[
        "accuracy_by_category"
    ][category] = {

        "total": total,

        "correct": corr,

        "accuracy_percent":
            round(
                corr / total * 100,
                4
            )
            if total
            else 0
    }


with open(
    SUMMARY_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n" + "=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print(
    f"Accuracy: {accuracy:.2f}%"
)

print(
    "\nSummary saved to:"
)

print(
    SUMMARY_FILE
)

print("=" * 70)
```
