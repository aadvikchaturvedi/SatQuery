```python
# ============================================================
# SATQUERY — TEST YOUR UPLOADED SATELLITE IMAGE
# ============================================================

import torch

def ask_satquery(question, max_new_tokens=60):

    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": image
                },
                {
                    "type": "text",
                    "text": (
                        "Answer the following question about the "
                        "satellite image concisely.\n\n"
                        f"Question: {question}"
                    )
                }
            ]
        }
    ]

    # Create Qwen vision-language prompt
    prompt = processor.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    # Process image + question
    inputs = processor(
        text=[prompt],
        images=[image],
        padding=True,
        return_tensors="pt"
    )

    # Move tensors to GPU
    inputs = {
        k: v.to(model.device) if torch.is_tensor(v) else v
        for k, v in inputs.items()
    }

    # Generate answer
    with torch.no_grad():

        output_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False
        )

    # Remove input tokens
    input_length = inputs["input_ids"].shape[1]

    generated_ids = output_ids[:, input_length:]

    answer = processor.batch_decode(
        generated_ids,
        skip_special_tokens=True
    )[0]

    return answer.strip()


# ============================================================
# 4 TEST QUESTIONS
# ============================================================

questions = [

    "Is there any visible water body in this satellite image?",

    "Does this satellite image contain urban or built-up areas?",

    "Is the area mainly covered by vegetation or agricultural land?",

    "What are the main land cover or geographic features visible in this satellite image?"
]


# ============================================================
# RUN TEST
# ============================================================

print("=" * 70)
print("SATQUERY — REAL-WORLD SATELLITE IMAGE TEST")
print("=" * 70)

for i, question in enumerate(questions, 1):

    print(f"\n{'=' * 70}")
    print(f"QUESTION {i}")
    print("=" * 70)

    print(question)

    answer = ask_satquery(question)

    print("\nMODEL ANSWER:")
    print(answer)

print("\n" + "=" * 70)
print("✅ ALL 4 TESTS COMPLETED")
print("=" * 70)
```
